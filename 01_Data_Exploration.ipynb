Import libraries

```python
import wfdb
import numpy as np
from sklearn.preprocessing import MinMaxScaler
```

Load ONE BIDMC record (training data)

```python
import wfdb
import numpy as np

record = wfdb.rdrecord(r"E:\open_data\bidmc01")

signal_names = [s.replace(",", "") for s in record.sig_name]

idx = {
    "ECG": signal_names.index("II"),
    "RESP": signal_names.index("RESP"),
    "PLETH": signal_names.index("PLETH")
}

data = np.stack([
    record.p_signal[:, idx["ECG"]],
    record.p_signal[:, idx["RESP"]],
    record.p_signal[:, idx["PLETH"]],
], axis=1)

data.shape
```

Normalize data (THIS CREATES data_scaled)

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
data_scaled = scaler.fit_transform(data)
```

create sliding windows (THIS CREATES X)

```python
def create_windows(data, window_size=100, step=50):
    windows = []
    for i in range(0, len(data) - window_size, step):
        windows.append(data[i:i + window_size])
    return np.array(windows)

X = create_windows(data_scaled, window_size=100, step=50)
X.shape
```

```python
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, LSTM, RepeatVector, TimeDistributed, Dense
```

Define the LSTM Autoencoder

```python
timesteps = X.shape[1]   # 100
features = X.shape[2]    # 3

inputs = Input(shape=(timesteps, features))

# Encoder
encoded = LSTM(64, activation="relu", return_sequences=False)(inputs)

# Bottleneck
encoded = RepeatVector(timesteps)(encoded)

# Decoder
decoded = LSTM(64, activation="relu", return_sequences=True)(encoded)
decoded = TimeDistributed(Dense(features))(decoded)

model = Model(inputs, decoded)
model.compile(optimizer="adam", loss="mse")

model.summary()
```

Train

```python
history = model.fit(
    X, X,
    epochs=10,
    batch_size=128,
    validation_split=0.1,
    verbose=1
)
```

Plot training loss

```python
import matplotlib.pyplot as plt

plt.plot(history.history["loss"], label="Train Loss")
plt.plot(history.history["val_loss"], label="Val Loss")
plt.legend()
plt.title("LSTM Autoencoder Training Loss")
plt.xlabel("Epoch")
plt.ylabel("MSE")
plt.show()
```

Reconstruction error

```python
X_pred = model.predict(X)

reconstruction_error = np.mean(
    np.square(X - X_pred),
    axis=(1, 2)
)
```

Threshold & anomaly flags

```python
threshold = np.mean(reconstruction_error) + 2 * np.std(reconstruction_error)
anomalies = reconstruction_error > threshold

print("Total windows:", len(anomalies))
print("Anomalies detected:", anomalies.sum())
```

Plot anomaly scores

```python
plt.figure(figsize=(12,4))
plt.plot(reconstruction_error, label="Reconstruction Error")
plt.axhline(threshold, color="red", linestyle="--", label="Threshold")
plt.legend()
plt.title("Anomaly Detection using LSTM Autoencoder")
plt.xlabel("Window Index")
plt.ylabel("Error")
plt.show()
```

GRU AUTOENCODER

```python
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, GRU, RepeatVector, TimeDistributed, Dense

inputs = Input(shape=(X.shape[1], X.shape[2]))

encoded = GRU(64, activation="relu")(inputs)
encoded = RepeatVector(X.shape[1])(encoded)

decoded = GRU(64, activation="relu", return_sequences=True)(encoded)
decoded = TimeDistributed(Dense(X.shape[2]))(decoded)

gru_model = Model(inputs, decoded)
gru_model.compile(optimizer="adam", loss="mse")

gru_history = gru_model.fit(
    X, X,
    epochs=10,
    batch_size=128,
    validation_split=0.1,
    verbose=1
)
```

CNN + LSTM (ADVANCED MODEL)

```python
from tensorflow.keras.layers import Conv1D, MaxPooling1D, Input, LSTM, RepeatVector, TimeDistributed, Dense
from tensorflow.keras.models import Model

inputs = Input(shape=(X.shape[1], X.shape[2]))

x = Conv1D(32, 3, activation="relu", padding="same")(inputs)
x = MaxPooling1D(2)(x)

x = LSTM(64, activation="relu")(x)
# Change this line to repeat the vector to match the original input length
x = RepeatVector(X.shape[1])(x)  # Use full length instead of X.shape[1]//2

x = LSTM(64, activation="relu", return_sequences=True)(x)
x = TimeDistributed(Dense(X.shape[2]))(x)

cnn_lstm_model = Model(inputs, x)
cnn_lstm_model.compile(optimizer="adam", loss="mse")

cnn_lstm_history = cnn_lstm_model.fit(
    X, X,
    epochs=10,
    batch_size=128,
    validation_split=0.1,
    verbose=1
)
```

LSTM vs GRU vs CNN-LSTM COMPARISON

```python
def reconstruction_error(model, X):
    X_pred = model.predict(X)
    return np.mean(np.square(X - X_pred), axis=(1,2))
```

```python
lstm_err = reconstruction_error(model, X)
gru_err = reconstruction_error(gru_model, X)
cnn_lstm_err = reconstruction_error(cnn_lstm_model, X)
```

Comparison Plot

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,4))
plt.plot(lstm_err, label="LSTM")
plt.plot(gru_err, label="GRU")
plt.plot(cnn_lstm_err, label="CNN-LSTM")
plt.legend()
plt.title("Model Comparison: Reconstruction Error")
plt.xlabel("Window Index")
plt.ylabel("Error")
plt.show()
```

Comparison Table

```python
import pandas as pd

results = pd.DataFrame({
    "Model": ["LSTM", "GRU", "CNN-LSTM"],
    "Mean Error": [lstm_err.mean(), gru_err.mean(), cnn_lstm_err.mean()],
    "Std Error": [lstm_err.std(), gru_err.std(), cnn_lstm_err.std()]
})

results
```
